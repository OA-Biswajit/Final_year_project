```python
from datasets import load_dataset

dataset = load_dataset(
    "ai4bharat/indic-align",
    "Wiki_Chat",
    split="train"
)

print(dataset)
```

```python
columns_needed = [
    "eng_Latn",
    "hin_Deva",
    "ory_Orya",
    "hin_Latn",
    "ory_Latn"
]

dataset_filtered = dataset.remove_columns(
    [col for col in dataset.column_names if col not in columns_needed]
)

print(dataset_filtered)
```

```python
import pandas as pd

df = dataset_filtered.to_pandas()

print(df.head())
```

```python
df.to_csv("Wiki_Chat_extract.csv", index=False)
```
