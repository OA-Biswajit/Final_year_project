```python
import pandas as pd
import os

print('✅ Ready!')
```

```python
CSV_FILES = [
    '/content/Dolly_T_extract_odia_hindi_pairs.csv',
    '/content/OpenAssistant_T_odia_hindi_pairs.csv',
    '/content/WikiHow_extract_odia_hindi_pairs.csv',
    '/content/hindi_odia_translation_dataset.csv',
]

OUTPUT_FILE = '/content/merged_odia_hindi.csv'


all_dfs = []

for path in CSV_FILES:
    if not os.path.exists(path):
        print(f'❌ File not found: {path}  — skipping')
        continue

    df = pd.read_csv(path, encoding='utf-8-sig')
    print(f'\n📄 {os.path.basename(path)}')
    print(f'   Columns : {list(df.columns)}')
    print(f'   Rows    : {len(df):,}')

    # Check both required columns exist
    missing = [c for c in ['ory_Orya', 'hin_Deva'] if c not in df.columns]
    if missing:
        print(f'   ❌ Missing columns: {missing} — skipping this file')
        continue

    df_pair = df[['ory_Orya', 'hin_Deva']].copy()

    df_pair = df_pair[
        df_pair['ory_Orya'].astype(str).str.strip().astype(bool) &
        df_pair['hin_Deva'].astype(str).str.strip().astype(bool)
    ]

    print(f'   Valid pairs : {len(df_pair):,}')
    all_dfs.append(df_pair)

if not all_dfs:
    print('\n❌ No valid data found. Check your file paths and column names.')
else:
    merged = pd.concat(all_dfs, ignore_index=True)

    before_dedup = len(merged)
    merged = merged.drop_duplicates()
    after_dedup = len(merged)

    merged.to_csv(OUTPUT_FILE, index=False, encoding='utf-8-sig')

    print(f'\n──────────────────────────────────────────')
    print(f'✅ Done!')
    print(f'   Total rows before dedup : {before_dedup:,}')
    print(f'   Duplicate rows removed  : {before_dedup - after_dedup:,}')
    print(f'   Final row count         : {after_dedup:,}')
    print(f'   Saved to                : {OUTPUT_FILE}')
    print(f'\n💡 Download the file from the 📁 sidebar on the left')

    # Preview
    print('\n── Sample (first 3 rows) ───────────────────')
    for i, row in merged.head(3).iterrows():
        print(f"\nRow {i+1}:")
        print(f"  ory_Orya : {str(row['ory_Orya'])[:100]}")
        print(f"  hin_Deva : {str(row['hin_Deva'])[:100]}")
```
